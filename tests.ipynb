```python
import itertools
import numpy as np
import fourier_lattice as fl
import sdf_lattice as sdf
import instanced_mesh as rmsh
```

```python
N, rho = 96, 0.30
A = fl.cubic_cell()
segs = fl.octet_segments()
```

```python
k = 0.02
f = sdf.lattice_field(segs, N, A, smooth=k)
r = sdf.radius_for_density(-f, rho)
```

```python
rho_fun = lambda z: 0.5 + (0.1 - 0.5) * z  
verts, faces, _ = fl.stacked_lattice_mesh(A, f, rho=rho_fun, cells=(2, 2, 4))
fl.write_stl("test.stl", verts, faces)
```

```python
verts, faces = rmsh.instanced_stack_mesh(A, f, rho=rho, cells=(6,6,4), seam_spacing=2.5)
fl.write_stl("test_instanced.stl", verts, faces)
```
